```python
# Carga de datos desde un archivo TXT en un DataFrame 
archivo = '/Users/israeljuarezjimenez/Documents/BDA/practica_5/empleados.txt'

def read_file(file_path, encoding='utf-8'):
    try:
        with open(file_path, 'r', encoding=encoding) as file:
            content = file.read()
    except UnicodeDecodeError:
        with open(file_path, 'r', encoding='latin-1') as file:
            content = file.read()
    return content

content = read_file(archivo)

blocks = content.strip().split('\n\n')

def process_block(block):
    employee_data = {}
    for line in block.split('\n'):
        if ': ' in line:
            key, value = line.split(': ', 1)
            employee_data[key.strip()] = value.strip()
    return employee_data

data = [process_block(block) for block in blocks if block.strip()]

df_txt = pd.DataFrame(data)

print(df_txt)
```

```python
# Carga de datos desde un archivo SQL en un DataFrame 
# Establecer conexión a la base de datos MySQL
bd_nombre = 'Empleados_bd'
usuario = 'root'
contrasena = 'Juarez31'
host = 'localhost'
puerto = '3306'
cadena_conexion = f'mysql+pymysql://{usuario}:{contrasena}@{host}:{puerto}/{bd_nombre}'
engine = create_engine(cadena_conexion)

# Leer y ejecutar un script SQL desde un archivo
name_file = '/Users/israeljuarezjimenez/Documents/BDA/practica_5/empleados.sql'
with open(name_file, 'r', encoding='utf-8') as file:
    sql_script = file.read()

commands = sql_script.split(';')
with engine.begin() as connection:
    for command in commands:
        if command.strip():
            connection.execute(text(command))

# Cargar datos de tablas específicas en DataFrames y mostrarlos
tablas = ['departamento', 'empleado', 'proyecto', 'empleado_proyecto']
dataframes = {}
for tabla in tablas:
    query = f"SELECT * FROM {tabla};"
    df = pd.read_sql_query(query, engine)
    dataframes[tabla] = df
    print(df)
    print("\n")

# Mostrar el DataFrame de empleados específicamente
df_sql = dataframes['empleado']
print("DataFrame de Empleados:")
print(df_sql)
```

```python
# Renombrar columnas en DataFrame cargado desde CSV para estandarizar los nombres de las columnas
df_csv = df_csv.rename(columns={'empno': 'Empleado', 'nombre': 'Nombre', 'puesto': 'Puesto', 
                                'jefe': 'Jefe', 'salario': 'Salario', 'comision': 'Comisión',
                                'departamento': 'Departamento'})
# Renombrar columnas en DataFrame cargado desde JSON para estandarizar los nombres de las columnas
df_json = df_json.rename(columns={'empno': 'Empleado', 'nombre': 'Nombre', 'jefe': 'JefeNo', 
                                  'fecha_contratacion': 'Fecha_Contratacion', 'salario': 'Salario', 
                                  'deptno': 'DeptNo'})
# Renombrar columnas en DataFrame cargado desde TXT para estandarizar los nombres de las columnas
df_txt = df_txt.rename(columns={'empleado': 'Nombre', 'sueldo': 'Salario', 'fecha de contratación': 'Fecha_Contratacion',
                                'departamento': 'Departamento', 'comisiones': 'Comisión'})
# Renombrar columnas en DataFrame cargado desde SQL para estandarizar los nombres de las columnas
df_sql = df_sql.rename(columns={'empno': 'Empleado', 'nombre': 'Nombre', 'puesto': 'Puesto', 'jefe': 'JefeNo',
                                'fecha_contratacion': 'Fecha_Contratacion', 'salario': 'Salario', 
                                'comision': 'Comisión', 'deptno': 'DeptNo'})
print('----------------------------------------------------------------------------')
print(df_csv)
print('----------------------------------------------------------------------------')
print(df_json)
print('----------------------------------------------------------------------------')
print(df_txt)
print('----------------------------------------------------------------------------')
print(df_sql)
```

```python
# Concatenar múltiples DataFrames en uno solo
# Los DataFrames df_sql, df_json, df_txt, y df_csv se combinan en df_general
df_general = pd.concat([df_sql,df_json,df_txt,df_csv], ignore_index=True)
```

```python
df_general
```

```python
# Limpieza de datos: Rellenar valores faltantes
df_general['Empleado'] = df_general['Empleado'].ffill() 
df_general['Jefe'] = df_general.groupby('Empleado')['Jefe'].ffill().bfill()
df_general['Salario'] = df_general.groupby('Empleado')['Salario'].ffill().bfill()
df_general['Comisión'] = df_general.groupby('Empleado')['Comisión'].ffill().bfill()
df_general['Departamento'] = df_general.groupby('Empleado')['Departamento'].ffill().bfill()
df_general['JefeNo'] = df_general.groupby('Empleado')['JefeNo'].ffill().bfill()
df_general['DeptNo'] = df_general.groupby('Empleado')['DeptNo'].ffill().bfill()

# Relleno de valores faltantes para la fecha de contratación
df_general['Fecha_Contratacion'] = df_general.groupby('Empleado')['Fecha_Contratacion'].ffill().bfill()

# Transformación: Estandarización de formatos de fecha (Si se reintroduce 'Fecha_Contratacion')
df_general['Fecha_Contratacion'] = pd.to_datetime(df_general['Fecha_Contratacion'], errors='coerce')

# Transformación: Eliminación de registros duplicados, conservando el último registro de cada empleado
df_general = df_general.drop_duplicates(subset='Empleado', keep='last')

# Mostrar las primeras filas del DataFrame limpio
df_general
```

```python
# Función para consolidar información en strings
def join_unique(values):
    unique_vals = set(values.dropna())
    return ', '.join(map(str, unique_vals))

# Agrupar por 'Empleado' y agregar datos
df_final = df_general.groupby('Empleado').agg({
    'Nombre': 'first',  
    'Puesto': 'first', 
    'JefeNo': 'first',  
    'Fecha_Contratacion': 'first', 
    'Salario': 'max',  
    'Comisión': 'sum',  
    'DeptNo': 'first',  
    'Departamento': join_unique, 
    'Jefe': join_unique 
}).reset_index()

df_final
```

```python
# Carga de datos limpios y transformados a la base de datos
df_final.to_sql(name='data_frame_final', con=engine, if_exists='replace', index=False)
engine.dispose() 
```
